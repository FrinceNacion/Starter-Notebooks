## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
from pathlib import Path

repo_root = Path.cwd()

csv_path = repo_root / "data" / "raw" / "content_refresh_anonymized.csv"
df = pd.read_csv(csv_path)

summary = {
    "search_volume": [df["search_volume"].mean(), df["search_volume"].median(), df["search_volume"].max()],
    "avg_position": [df["avg_position"].mean(), df["avg_position"].median(), df["avg_position"].max()],
    "sessions_last_30d": [df["sessions_last_30d"].mean(), df["sessions_last_30d"].median(), df["sessions_last_30d"].max()],
    "content_age_days": [df["content_age_days"].mean(), df["content_age_days"].median(), df["content_age_days"].max()]
}

for metric, vals in summary.items():
    print(f"{metric}: mean={vals[0]:.2f}, median={vals[1]:.2f}, max={vals[2]:.2f}")
```
